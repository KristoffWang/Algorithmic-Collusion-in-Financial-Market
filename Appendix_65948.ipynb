```python
#Section 4.2

import numpy as np
from scipy.optimize import brentq


I = 2
sigma_v = 1.0
sigma_u = 0.1
theta = 0.1
xi_list = [0, 50, 500]  

def nash_eq(lam, I, sigma_v, sigma_u, theta, xi):
    chi = 1 / ((I + 1) * lam)
    gamma = (I * chi) / ((I * chi)**2 + (sigma_u / sigma_v)**2)
    lam_from_pricing = (theta * gamma + xi) / (theta + xi**2)
    return lam - lam_from_pricing  


def cartel_eq(lam, I, sigma_v, sigma_u, theta, xi):
    chi = 1 / (2 * I * lam)
    gamma = (I * chi) / ((I * chi)**2 + (sigma_u / sigma_v)**2)
    lam_from_pricing = (theta * gamma + xi) / (theta + xi**2)
    return lam - lam_from_pricing


results = []
for xi in xi_list:
    lambda_N = brentq(nash_eq, 1e-6, 10, args=(I, sigma_v, sigma_u, theta, xi))
    chi_N = 1 / ((I + 1) * lambda_N)
    gamma_N = (I * chi_N) / ((I * chi_N)**2 + (sigma_u / sigma_v)**2)
    pi_N = (1 - lambda_N * I * chi_N) * chi_N * sigma_v**2
    lambda_M = brentq(cartel_eq, 1e-6, 10, args=(I, sigma_v, sigma_u, theta, xi))
    chi_M = 1 / (2 * I * lambda_M)
    gamma_M = (I * chi_M) / ((I * chi_M)**2 + (sigma_u / sigma_v)**2)
    pi_M = (1 - lambda_M * I * chi_M) * chi_M * sigma_v**2

    results.append([xi, chi_N, lambda_N, pi_N, chi_M, lambda_M, pi_M])


print("="*90)
print(f"{'ξ':<6} {'χ^N':<10} {'λ^N':<10} {'π^N':<10} {'χ^M':<10} {'λ^M':<10} {'π^M':<10}")
print("="*90)
for res in results:
    xi, chi_N, lambda_N, pi_N, chi_M, lambda_M, pi_M = res
    print(f"{xi:<6} {chi_N:<10.6f} {lambda_N:<10.6f} {pi_N:<10.6f} {chi_M:<10.6f} {lambda_M:<10.6f} {pi_M:<10.6f}")
```

```python
#Section 4.4


import numpy as np
from scipy.optimize import brentq

sigma_v = 1.0
sigma_u = 0.1
theta = 0.1
xi = 500
I_list = [2, 3, 5]

def lambda_nash(I_star):
    def f(lam):
        chi = 1 / ((I_star + 1) * lam)
        gamma = (I_star * chi) / ((I_star * chi)**2 + (sigma_u / sigma_v)**2)
        lam_implied = (theta * gamma + xi) / (theta + xi**2)
        return lam - lam_implied
    return brentq(f, 1e-8, 10)

def lambda_cartel(I):
    def f(lam):
        chi = 1 / (2 * I * lam)
        gamma = (I * chi) / ((I * chi)**2 + (sigma_u / sigma_v)**2)
        lam_implied = (theta * gamma + xi) / (theta + xi**2)
        return lam - lam_implied
    return brentq(f, 1e-8, 10)

def solve_I_star(I):
    target_lam = lambda_cartel(I)
    def f(I_star):
        return lambda_nash(I_star) - target_lam
    I_star = brentq(f, 1e-6, 100)
    return I_star, target_lam

print("="*60)
print(f"{'I':<4} {'λ^M(I)':<12} {'I^*':<12} {'I^*=1?':<10}")
print("="*60)
for I in I_list:
    I_star, lam_M = solve_I_star(I)
    is_one = np.isclose(I_star, 1, atol=1e-3)
    print(f"{I:<4} {lam_M:<12.8f} {I_star:<12.4f} {is_one:<10}")
```

```python
#Section 5

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq
from scipy.stats import norm
from numba import jit
import pickle

V_BAR = 1.0
SIGMA_V = 1.0
SIGMA_U = 0.1
I = 2
RHO = 0.95
THETA = 0.1
ALPHA = 0.05
BETA = 5e-6
IOTA = 0.1
N_V = 10
N_X = 15
N_P = 31
np.random.seed(520)

def solve_equilibrium(xi, I_spec=2):
    def nash_f(lam):
        chi = 1 / ((I_spec + 1) * lam)
        gamma = (I_spec * chi) / ((I_spec * chi)**2 + (SIGMA_U / SIGMA_V)**2)
        lam_implied = (THETA * gamma + xi) / (THETA + xi**2)
        return lam - lam_implied
    
    def cartel_f(lam):
        chi = 1 / (2 * I_spec * lam)
        gamma = (I_spec * chi) / ((I_spec * chi)**2 + (SIGMA_U / SIGMA_V)**2)
        lam_implied = (THETA * gamma + xi) / (THETA + xi**2)
        return lam - lam_implied
    
    lam_N = brentq(nash_f, 1e-8, 10)
    chi_N = 1 / ((I_spec + 1) * lam_N)
    lam_M = brentq(cartel_f, 1e-8, 10)
    chi_M = 1 / (2 * I_spec * lam_M)
    
    return chi_N, chi_M, lam_N, lam_M

def build_grids(chi_N, chi_M, lam_N, I_spec=2):
    v_quantiles = (2 * np.arange(1, N_V+1) - 1) / (2 * N_V)
    v_grid = V_BAR + SIGMA_V * norm.ppf(v_quantiles)
    
    action_grids = []
    price_grids = []
    x_N_list = []
    x_M_list = []
    
    for v_k in v_grid:
        delta_v = v_k - V_BAR
        x_N_k = chi_N * delta_v
        x_M_k = chi_M * delta_v
        x_N_list.append(x_N_k)
        x_M_list.append(x_M_k)
        
        delta_x = x_N_k - x_M_k
        if delta_v > 0:
            x_low = x_M_k - IOTA * delta_x
            x_high = x_N_k + IOTA * delta_x
        else:
            x_low = x_M_k + IOTA * delta_x
            x_high = x_N_k - IOTA * delta_x
        action_grid_k = np.linspace(x_low, x_high, N_X)
        action_grids.append(action_grid_k)
        
        max_x = max(x_N_k, x_M_k)
        min_x = min(x_N_k, x_M_k)
        p_Hk = V_BAR + lam_N * (I_spec * max_x + 1.96 * SIGMA_U)
        p_Lk = V_BAR + lam_N * (I_spec * min_x - 1.96 * SIGMA_U)
        p_range = p_Hk - p_Lk
        p_low = p_Lk - IOTA * p_range
        p_high = p_Hk + IOTA * p_range
        price_grid_k = np.linspace(p_low, p_high, N_P)
        price_grids.append(price_grid_k)
    
    return v_grid, np.array(action_grids), np.array(price_grids), np.array(x_N_list), np.array(x_M_list)


@jit(nopython=True)
def encode_state(p_prev_idx, v_prev_idx, v_curr_idx, n_p, n_v):
    return p_prev_idx * n_v * n_v + v_prev_idx * n_v + v_curr_idx

@jit(nopython=True)
def q_learning_core_loop(T, Q, v_grid, action_grids, price_grids, lam_N, I_spec):
    n_p = N_P
    n_v = N_V
    n_x = N_X
    
    t_v_count = np.zeros(n_v, dtype=np.int64)
    v_prev_idx = np.random.randint(0, n_v)
    v_curr_idx = np.random.randint(0, n_v)
    p_prev_idx = np.random.randint(0, n_p)
    
    record_interval = T // 1000
    policy_history = np.zeros((record_interval, n_v), dtype=np.int64)
    history_ptr = 0
    
    for t in range(T):
        state_idx = encode_state(p_prev_idx, v_prev_idx, v_curr_idx, n_p, n_v)
        v_curr = v_grid[v_curr_idx]
        action_grid_curr = action_grids[v_curr_idx]
        price_grid_curr = price_grids[v_curr_idx]
        t_v_count[v_curr_idx] += 1
        epsilon = np.exp(-BETA * t_v_count[v_curr_idx])
        if np.random.random() < epsilon:
            x_idx = np.random.randint(0, n_x)
        else:
            x_idx = np.argmax(Q[state_idx, :])
        x_i = action_grid_curr[x_idx]
        
        total_x = I_spec * x_i
        u_t = np.random.normal(0, SIGMA_U)
        y_t = total_x + u_t
        p_t = V_BAR + lam_N * y_t
        pi_t = (v_curr - p_t) * x_i
        
        v_next_idx = np.random.randint(0, n_v)
        p_t_idx = np.argmin(np.abs(price_grid_curr - p_t))
        next_state_idx = encode_state(p_t_idx, v_curr_idx, v_next_idx, n_p, n_v)
        
        max_q_next = np.max(Q[next_state_idx, :])
        Q[state_idx, x_idx] = (1 - ALPHA) * Q[state_idx, x_idx] + ALPHA * (pi_t + RHO * max_q_next)
        
        p_prev_idx = p_t_idx
        v_prev_idx = v_curr_idx
        v_curr_idx = v_next_idx
        
        if t % (T // record_interval) == 0 and history_ptr < record_interval:
            for v_idx in range(n_v):
                action_count = np.zeros(n_x, dtype=np.int64)
                for v_prev in range(n_v):
                    for p_prev in range(n_p):
                        temp_state_idx = encode_state(p_prev, v_prev, v_idx, n_p, n_v)
                        best_x_idx = np.argmax(Q[temp_state_idx, :])
                        action_count[best_x_idx] += 1
                policy_history[history_ptr, v_idx] = np.argmax(action_count)
            history_ptr += 1
    
    return Q, t_v_count, policy_history


def q_learning_simulation(T, v_grid, action_grids, price_grids, lam_N, chi_N, chi_M, I_spec=2):
    n_p = N_P
    n_v = N_V
    n_x = N_X
    n_states = n_p * n_v * n_v
    Q = np.zeros((n_states, n_x), dtype=np.float64)
    for v_curr_idx in range(n_v):
        v_k = v_grid[v_curr_idx]
        action_grid_k = action_grids[v_curr_idx]
        for x_n_idx in range(n_x):
            x_n = action_grid_k[x_n_idx]
            sum_payoff = 0.0
            if I_spec == 1:
                price = V_BAR + lam_N * x_n
                sum_payoff = (v_k - price) * x_n
            else:
                for x_other in action_grid_k:
                    price = V_BAR + lam_N * (x_n + (I_spec-1)*x_other)
                    payoff = (v_k - price) * x_n
                    sum_payoff += payoff
            q0 = sum_payoff / ((1 - RHO) * n_x)
            noise_std = np.abs(q0) * 0.001
            for v_prev_idx in range(n_v):
                for p_prev_idx in range(n_p):
                    state_idx = encode_state(p_prev_idx, v_prev_idx, v_curr_idx, n_p, n_v)
                    Q[state_idx, x_n_idx] = q0 + np.random.normal(0, noise_std)
    
    Q, t_v_count, policy_history = q_learning_core_loop(T, Q, v_grid, action_grids, price_grids, lam_N, I_spec)
    
    converged_start = int(0.9 * len(policy_history))
    converged_policy = policy_history[converged_start:]

    chi_C_list = []
    avg_abs_x = 0.0
    for v_idx in range(n_v):
        state_idx = encode_state(0, 0, v_idx, n_p, n_v)
        best_x_idx = np.argmax(Q[state_idx, :])
        best_x = action_grids[v_idx, best_x_idx]
        avg_abs_x += np.abs(best_x)
    avg_abs_x /= n_v

    for policy_step in converged_policy:
        for v_idx in range(n_v):
            best_x_idx = policy_step[v_idx]
            best_x = action_grids[v_idx, best_x_idx]
            delta_v = v_grid[v_idx] - V_BAR
            if np.abs(delta_v) > 1e-6:
                chi_C_list.append(best_x / delta_v)

    chi_C = np.mean(chi_C_list)
    denominator = chi_N - chi_M
    delta_chi = 0.0
    if np.abs(denominator) > 1e-6:
        delta_chi = (chi_N - chi_C) / denominator

    record_interval = len(policy_history)
    delta_chi_history = np.zeros(record_interval)
    if np.abs(denominator) > 1e-6:
        for i in range(record_interval):
            temp_chi_list = []
            for v_idx in range(n_v):
                best_x_idx = policy_history[i, v_idx]
                best_x = action_grids[v_idx, best_x_idx]
                delta_v = v_grid[v_idx] - V_BAR
                if np.abs(delta_v) > 1e-6:
                    temp_chi_list.append(best_x / delta_v)
            if len(temp_chi_list) > 0:
                temp_chi = np.mean(temp_chi_list)
                delta_chi_history[i] = (chi_N - temp_chi) / denominator
    
    return Q, chi_C, delta_chi, avg_abs_x, policy_history, delta_chi_history


def run_irf(Q, v_grid, action_grids, price_grids, lam_N, avg_abs_x, shock_size_pct, I_spec=2, n_sim=10000, H=9):
    n_v = N_V
    n_p = N_P
    shock_t = 3
    u_shock = shock_size_pct * avg_abs_x

    base_p_all = np.zeros((n_sim, H+1))
    base_x_all = np.zeros((n_sim, H+1))
    base_pi_all = np.zeros((n_sim, H+1))
    shock_p_all = np.zeros((n_sim, H+1))
    shock_x_all = np.zeros((n_sim, H+1))
    shock_pi_all = np.zeros((n_sim, H+1))

    for sim in range(n_sim):
        v_seq = np.random.randint(0, n_v, H+2)
        u_seq = np.random.normal(0, SIGMA_U, H+1)

        p_prev_idx_base = 0
        v_prev_idx_base = v_seq[0]
        for t in range(1, H+1):
            v_curr_idx_base = v_seq[t]
            state_idx_base = encode_state(p_prev_idx_base, v_prev_idx_base, v_curr_idx_base, n_p, n_v)
            x_idx_base = np.argmax(Q[state_idx_base, :])
            x_base = action_grids[v_curr_idx_base, x_idx_base]
            u_t_base = u_seq[t-1]

            total_x_base = I_spec * x_base
            y_t_base = total_x_base + u_t_base
            p_t_base = V_BAR + lam_N * y_t_base
            pi_t_base = (v_grid[v_curr_idx_base] - p_t_base) * x_base

            delta_v_base = v_grid[v_curr_idx_base] - V_BAR
            sign_base = np.sign(delta_v_base) if delta_v_base != 0 else 1
            p_tilde_base = (p_t_base - V_BAR) * sign_base
            x_tilde_base = x_base * sign_base

            base_p_all[sim, t] = p_tilde_base
            base_x_all[sim, t] = x_tilde_base
            base_pi_all[sim, t] = pi_t_base

            p_t_idx_base = np.argmin(np.abs(price_grids[v_curr_idx_base] - p_t_base))
            p_prev_idx_base = p_t_idx_base
            v_prev_idx_base = v_curr_idx_base

        
        p_prev_idx_shock = 0
        v_prev_idx_shock = v_seq[0]
        for t in range(1, H+1):
            v_curr_idx_shock = v_seq[t]
            state_idx_shock = encode_state(p_prev_idx_shock, v_prev_idx_shock, v_curr_idx_shock, n_p, n_v)
            x_idx_shock = np.argmax(Q[state_idx_shock, :])
            x_shock = action_grids[v_curr_idx_shock, x_idx_shock]
            u_t_shock = u_seq[t-1]

            if t == shock_t:
                delta_v_shock = v_grid[v_curr_idx_shock] - V_BAR
                sign_shock = np.sign(delta_v_shock) if delta_v_shock != 0 else 1
                u_t_shock += u_shock * sign_shock

            total_x_shock = I_spec * x_shock
            y_t_shock = total_x_shock + u_t_shock
            p_t_shock = V_BAR + lam_N * y_t_shock
            pi_t_shock = (v_grid[v_curr_idx_shock] - p_t_shock) * x_shock

            delta_v_shock = v_grid[v_curr_idx_shock] - V_BAR
            sign_shock = np.sign(delta_v_shock) if delta_v_shock != 0 else 1
            p_tilde_shock = (p_t_shock - V_BAR) * sign_shock
            x_tilde_shock = x_shock * sign_shock

            shock_p_all[sim, t] = p_tilde_shock
            shock_x_all[sim, t] = x_tilde_shock
            shock_pi_all[sim, t] = pi_t_shock

            p_t_idx_shock = np.argmin(np.abs(price_grids[v_curr_idx_shock] - p_t_shock))
            p_prev_idx_shock = p_t_idx_shock
            v_prev_idx_shock = v_curr_idx_shock

    base_mean_p = np.mean(base_p_all, axis=0)
    base_mean_x = np.mean(base_x_all, axis=0)
    base_mean_pi = np.mean(base_pi_all, axis=0)
    shock_mean_p = np.mean(shock_p_all, axis=0)
    shock_mean_x = np.mean(shock_x_all, axis=0)
    shock_mean_pi = np.mean(shock_pi_all, axis=0)

    min_denominator = 1e-6
    price_dev = (shock_mean_p - base_mean_p) / (np.maximum(np.abs(base_mean_p), min_denominator)) * 100
    order_dev = (shock_mean_x - base_mean_x) / (np.maximum(np.abs(base_mean_x), min_denominator)) * 100
    profit_dev = (shock_mean_pi - base_mean_pi) / (np.maximum(np.abs(base_mean_pi), min_denominator)) * 100

    return price_dev, order_dev, profit_dev


if __name__ == "__main__":
    print("="*60)
    print("Running Baseline Scenario: ξ=0")
    print("="*60)
    chi_N_0, chi_M_0, lam_N_0, lam_M_0 = solve_equilibrium(xi=0, I_spec=2)
    print(f"ξ=0 Equilibrium Result：χ^N={chi_N_0:.6f}, χ^M={chi_M_0:.6f}, λ^N={lam_N_0:.6f}")
    v_grid_0, action_grids_0, price_grids_0, x_N_0, x_M_0 = build_grids(chi_N_0, chi_M_0, lam_N_0, I_spec=2)
    
    print("\n[Step 1] Running Sanity Check: I=1, ξ=0, T=10×10^6")
    chi_N_sanity, chi_M_sanity, lam_N_sanity, lam_M_sanity = solve_equilibrium(xi=0, I_spec=1)
    v_grid_sanity, action_grids_sanity, price_grids_sanity, _, _ = build_grids(chi_N_sanity, chi_M_sanity, lam_N_sanity, I_spec=1)
    Q_sanity, chi_C_sanity, delta_chi_sanity, _, _, _ = q_learning_simulation(
        T=10_000_000,
        v_grid=v_grid_sanity,
        action_grids=action_grids_sanity,
        price_grids=price_grids_sanity,
        lam_N=lam_N_sanity,
        chi_N=chi_N_sanity,
        chi_M=chi_M_sanity,
        I_spec=1
    )
    print(f"Sanity Check Result：")
    print(f"Analytical Nash Value χ^N={chi_N_sanity:.6f}, Learned Trading Intensity χ^C={chi_C_sanity:.6f}")
    print(f"Relative Error：{np.abs(chi_C_sanity - chi_N_sanity)/chi_N_sanity*100:.4f}%")
    
    print("\n[Step 2] Running Baseline Simulation: I=2, ξ=0, T=50×10^6")
    Q_0, chi_C_0, delta_chi_0, avg_abs_x_0, policy_history_0, delta_chi_history_0 = q_learning_simulation(
        T=50_000_000,
        v_grid=v_grid_0,
        action_grids=action_grids_0,
        price_grids=price_grids_0,
        lam_N=lam_N_0,
        chi_N=chi_N_0,
        chi_M=chi_M_0,
        I_spec=2
    )
    print(f"ξ=0 Simulation Result：")
    print(f"Learned Trading Intensity χ^C={chi_C_0:.6f}, Collusion Index Δ_χ={delta_chi_0:.4f}")
    with open("q_learning_xi0.pkl", "wb") as f:
        pickle.dump({
            "Q": Q_0, "chi_C": chi_C_0, "delta_chi": delta_chi_0,
            "avg_abs_x": avg_abs_x_0, "policy_history": policy_history_0,
            "delta_chi_history": delta_chi_history_0
        }, f)
    
    print("\n" + "="*60)
    print("Running Scenario: ξ=500")
    print("="*60)
    chi_N_500, chi_M_500, lam_N_500, lam_M_500 = solve_equilibrium(xi=500, I_spec=2)
    print(f"ξ=500 Equilibrium Result：χ^N={chi_N_500:.6f}, χ^M={chi_M_500:.6f}, λ^N={lam_N_500:.6f}")
    v_grid_500, action_grids_500, price_grids_500, x_N_500, x_M_500 = build_grids(chi_N_500, chi_M_500, lam_N_500, I_spec=2)
    
    print("\n[Step 3] Running Simulation: I=2, ξ=500, T=500×10^6")
    Q_500, chi_C_500, delta_chi_500, avg_abs_x_500, policy_history_500, delta_chi_history_500 = q_learning_simulation(
        T=500_000_000,
        v_grid=v_grid_500,
        action_grids=action_grids_500,
        price_grids=price_grids_500,
        lam_N=lam_N_500,
        chi_N=chi_N_500,
        chi_M=chi_M_500,
        I_spec=2
    )
    print(f"ξ=500 Simulation Result：")
    print(f"Learned Trading Intensity χ^C={chi_C_500:.6f}, Collusion Index Δ_χ={delta_chi_500:.4f}")
    with open("q_learning_xi500.pkl", "wb") as f:
        pickle.dump({
            "Q": Q_500, "chi_C": chi_C_500, "delta_chi": delta_chi_500,
            "avg_abs_x": avg_abs_x_500, "policy_history": policy_history_500,
            "delta_chi_history": delta_chi_history_500
        }, f)
    
    print("\n" + "="*60)
    print("Running Impulse Response Analysis")
    print("="*60)
    shock_sizes = [0.005, 0.025, 0.07]
    irf_results_0 = {}
    irf_results_500 = {}
    
    for s in shock_sizes:
        print(f"Running IRF for shock size: {s*100}%")
        p_dev_0, x_dev_0, pi_dev_0 = run_irf(
            Q_0, v_grid_0, action_grids_0, price_grids_0, lam_N_0, avg_abs_x_0, s, I_spec=2
        )
        irf_results_0[s] = (p_dev_0, x_dev_0, pi_dev_0)
        
        p_dev_500, x_dev_500, pi_dev_500 = run_irf(
            Q_500, v_grid_500, action_grids_500, price_grids_500, lam_N_500, avg_abs_x_500, s, I_spec=2
        )
        irf_results_500[s] = (p_dev_500, x_dev_500, pi_dev_500)
    
    with open("irf_results.pkl", "wb") as f:
        pickle.dump({"xi0": irf_results_0, "xi500": irf_results_500}, f)
    
    print("\n All simulations completed! All results are saved.")
```
